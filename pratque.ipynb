````python
# on recupere la colonne salary 

cible = df['Salary']
cible.describe()
````

## 2. les stats basiques et analyses

````python
moyenne = cible.mean() # moyenne
mediane = cible.median() # plus fiable
vmax = cible.max()
vmin = cible.min()
mode = cible.mode() # mode : la variable ayant pus des valeurs dans la population 
````

````python
print(f"la moyenne des salaires est de :{moyenne}")
print(f"la mediane (valeure du milieu) est de : {mediane}")
print(f"les modes sont : \n {mode}") # peux renvoyer pusieurs valeures
````

**interprétation**:
- si la moyenne est proche de la mediane la distribution est relativement symétrique
- si la moyenne est nettement < mediane -> asymetrie à droite
- si la moyenne est nettement > mediane -> asymetrie à gauche 

## 3. les mesures de dispersion

````python
variance = cible.var(ddof=1)
ecart_type = cible.std(ddof=1)
etendue =  cible.max() - cible.min()
cv = (ecart_type/moyenne) * 100

print(f"la variance est de {variance:.2f}")
print(f"l'eacrt type est de {ecart_type:.2f}")
print(f"le cv est de {cv:.2f}")
print(f"l'étendue est de {etendue}")
````

## 4. mesures de position

### les quartiles

##

- Q1 : (premier quartil ) : 25 pourcents en dessous
- Q2 : (mediane) : 50 pourcents de part  et d'autres
- Q3 : (troisieme quartil) : 75 pourcents au dessus

elles aident à comprendre où se concentre la majorité des nos données

````python
q1 = cible.quantile(0.25)
q2 = cible.quantile(0.5)
q3 = cible.quantile(0.75)

print(f"Q1 = {q1:.2f}")
print(f"Q2 = {q2:.2f}")
print(f"Q3 = {q3:.2f}")
````

#### interprétation

1. 75% des salaires sont en dessous de 56720.7 et 25%  au dessus
2. c'est la mediane 
3. 75% sont au dessus de 100544 et 25 en dessous
