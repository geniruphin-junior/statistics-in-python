## 1. Introduction

**la statistique** est une science qui fait la collecte,l'interpretation et l'analyses des données pour repondre à des questions bien precises

**concepts importants :**
1. l'individu : est une unité de la population, il peux etre une personne,un produit,...
2. La population : est un ensemble de plusieurs individus de meme nature,
3. l'échantillon : est une partie representative d'une population,
4. sont des caracteristiques à une population (datatest) ex: mode,moyenne,mediane,max...
